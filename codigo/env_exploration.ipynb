```python
import gymnasium as gym
import rware
env = gym.make("rware-tiny-2ag-v2")
```

```python
EPISODE_LENGTH = 5000
TOTAL_EPISODES = 30
N_AGENTS = 2
```

```python
rewards = [[0] * EPISODE_LENGTH for _ in range(N_AGENTS)]  # Initialize a list to store rewards for each agent
```

```python
for k in range(TOTAL_EPISODES):
    obs = env.reset()  # a tuple of observations
    for i in range(EPISODE_LENGTH):
        actions = env.action_space.sample()  # the action space can be sampled
        obs, reward, terminated, truncated, info = env.step(actions)
        env.render()  # render the environment
        for agent_idx in range(N_AGENTS):
            rewards[agent_idx][i] += (reward[agent_idx] - rewards[agent_idx][i]) / (k + 1)  # Store the reward for each agent
    print(f"Episode {k + 1}/{TOTAL_EPISODES} completed.")
```

```python
print("Average rewards for each agent over all episodes:")
for agent_idx in range(N_AGENTS):
    avg_reward = rewards[agent_idx]
    print(f"Agent {agent_idx}: {avg_reward}")
```
